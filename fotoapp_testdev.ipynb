```python
# Punto 1: redimensionado para red social
from skimage import data
from PIL import Image
import matplotlib.pyplot as plt

Image.fromarray(data.hubble_deep_field()).save("imagenes_prueba/foto_redimensionada.jpg")

foto = fotoapp.redimensionar("imagenes_prueba/foto_redimensionada.jpg", "Instagram")

print("Tamano final:", foto.size)
plt.imshow(foto)
plt.axis("off")
plt.title("Punto 1: Redimensionada")
plt.show()
```

```python
# Punto 2: ecualizacion de histograma
import numpy as np
from skimage import data
from PIL import Image

img = data.moon().astype(np.uint8)
oscura = (img * 0.15).astype(np.uint8)  # imagen muy oscura
clara = np.clip(img * 0.15 + 210, 0, 255).astype(np.uint8)  # imagen muy clara

fotoapp.ecualizar(Image.fromarray(oscura), "prueba_oscura")
fotoapp.ecualizar(Image.fromarray(clara), "prueba_clara")
```

```python
# Punto 3: filtros de Pillow
from skimage import data
from PIL import Image

img_gato = Image.fromarray(data.chelsea())

filtros = ["Blur", "Contour", "Detail", "Edge Enhance", "Edge Enhance More",
           "Emboss", "Find Edges", "Sharpen", "Smooth"]

for nombre in filtros:
    print(f"--- Aplicando filtro: {nombre} ---")
    fotoapp.aplicar_filtro(img_gato, nombre)
```

```python
# Punto 4: boceto por bordes
from skimage import data
from PIL import Image

img_astronauta = Image.fromarray(data.astronaut())

fotoapp.hacer_boceto(img_astronauta)
print("Boceto generado y guardado en 'fotos_mejoradas'.")
```
